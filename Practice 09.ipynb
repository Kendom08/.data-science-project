```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams['figure.dpi'] = 120
```

```python
df = sns.load_dataset('diamonds')
print("数据形状：", df.shape)
df.head()
```

```python
print("总销售额：$", round(df['price'].sum(), 2))
print("平均价格：$", round(df['price'].mean(), 2))
print("克拉数范围：", df['carat'].min(), "~", df['carat'].max())
df.describe()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df.sample(2000, random_state=42), x='carat', y='price', 
                alpha=0.4, hue='cut', palette='viridis')
plt.title('Diamond Carat vs Price', fontsize=14)
plt.xlabel('Carat (Weight)')
plt.ylabel('Price ($)')
plt.tight_layout()
plt.savefig('carat_vs_price.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='cut', y='price', 
            order=['Fair', 'Good', 'Very Good', 'Premium', 'Ideal'],
            hue='cut', palette='Blues_r', legend=False)
plt.title('Price Distribution by Diamond Cut', fontsize=14)
plt.xlabel('Cut Quality')
plt.ylabel('Price ($)')
plt.tight_layout()
plt.savefig('price_by_cut.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
color_counts = df['color'].value_counts().sort_index()
sns.barplot(x=color_counts.index, y=color_counts.values,
            hue=color_counts.index, palette='Reds_r', legend=False)
plt.title('Diamond Count by Color Grade', fontsize=14)
plt.xlabel('Color Grade (D = Best, J = Worst)')
plt.ylabel('Number of Diamonds')
plt.tight_layout()
plt.savefig('count_by_color.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='color', y='price',
            order=['D', 'E', 'F', 'G', 'H', 'I', 'J'],
            hue='color', palette='Greens_r', legend=False)
plt.title('Price Distribution by Diamond Color Grade', fontsize=14)
plt.xlabel('Color Grade (D = Best, J = Worst)')
plt.ylabel('Price ($)')
plt.tight_layout()
plt.savefig('price_by_color.png', dpi=150)
plt.show()
```
